# Ejercicio 1: Ingesta y Preparación del Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Generamos datos de los últimos 12 meses
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
np.random.seed(42)

datos_it = {
    'Mes': meses,
    'Sueldo_Promedio_USD': [1100, 1150, 1200, 1180, 1250, 1300, 1350, 1400, 1380, 1450, 1500, 1550],
    'Puestos_Abiertos': np.random.randint(50, 200, 12)
}

df_it = pd.DataFrame(datos_it)
df_it.to_csv('mercado_it_2025.csv', index=False)
print('Archivo mercado_it_2025.csv generado con éxito.')
```

---

# Ejercicio 2: Gráfico de Línea (Tendencia Temporal)

```python
df_it = pd.read_csv('mercado_it_2025.csv')
print('Archivo mercado_it_2025.csv leído correctamente.')
```

```python
plt.figure()
plt.plot(df_it['Mes'], df_it['Sueldo_Promedio_USD'],
         color='green',
         marker='o')
plt.title('Evolución del Sueldo Promedio USD - Mercado IT Mendoza 2025')
plt.xlabel('Mes')
plt.ylabel('Sueldo Promedio (USD)')
plt.grid(True)
plt.show()
```

---

# Ejercicio 3: Gráfico de Barras (Comparativa de Demanda)

```python
plt.figure()
plt.bar(df_it['Mes'], df_it['Puestos_Abiertos'],
        color='orange')
plt.title('Puestos Abiertos por Mes - Mercado IT Mendoza 2025')
plt.xlabel('Mes')
plt.ylabel('Puestos Abiertos')
plt.show()

mes_mayor_demanda = df_it.loc[df_it['Puestos_Abiertos'].idxmax(), 'Mes']
print(f'\n📊 Mes con mayor demanda de puestos: {mes_mayor_demanda}')
```

---

# Ejercicio 4: Personalización y Estilo

```python
plt.style.use('ggplot')
plt.figure(figsize=(12, 6))

plt.plot(df_it['Mes'], df_it['Sueldo_Promedio_USD'],
         color='green',
         marker='o',
         linewidth=2)

promedio_anual = df_it['Sueldo_Promedio_USD'].mean()
plt.axhline(y=promedio_anual, color='red', linestyle='--', label=f'Promedio: ${promedio_anual:.0f}')

plt.title('Evolución del Sueldo Promedio USD - Mercado IT Mendoza 2025', fontsize=14)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Sueldo Promedio (USD)', fontsize=12)
plt.legend()
plt.grid(True)
plt.show()
```
