# Ejercicio 1: Preparación del Entorno y el Índice Real

```python
import pandas as pd
import numpy as np

np.random.seed(123)
legajos_generados = ['TUP-' + str(i) for i in range(1001, 1021)]
carreras_disponibles = ['Programacion', 'Redes', 'Sistemas']
datos_academicos = {
    'Legajo': legajos_generados,
    'Carrera': np.random.choice(carreras_disponibles, 20),
    'Matematica': np.random.randint(2, 11, 20),
    'Bases_Datos': np.random.randint(2, 11, 20),
    'Algoritmos': np.random.randint(2, 11, 20)
}
df_generador = pd.DataFrame(datos_academicos)
df_generador.to_csv('notas_semestre.csv', index=False)
print('Archivo notas_semestre.csv generado con éxito.')
```

```python
df_notas = pd.read_csv('notas_semestre.csv')
print('DataFrame leído correctamente.')
print(f'Forma del DataFrame: {df_notas.shape}')
print(f'Columnas: {df_notas.columns.tolist()}')
```

```python
df_notas = df_notas.set_index('Legajo')
print('Índice actualizado a la columna Legajo.')
```

```python
print('Primeras 5 filas (índice ahora es Legajo):')
df_notas.head()
```

# Ejercicio 3: Operando al Paciente (loc)

```python
df_notas.loc['TUP-1005', 'Bases_Datos'] = 9
print('Nota de TUP-1005 en Bases_Datos actualizada a 9.')
```

```python
df_notas.loc['TUP-1018', 'Matematica'] = 7
print('Nota de TUP-1018 en Matematica actualizada a 7.')
```

```python
df_verificacion = df_notas.loc[['TUP-1005', 'TUP-1018']]
print('=== VERIFICACIÓN DE MODIFICACIONES ===')
df_verificacion
```

# Ejercicio 4: Integración - Máscaras Booleanas Múltiples

```python
mascara = (
    (df_notas['Carrera'] == 'Programacion') &
    (df_notas['Algoritmos'] >= 8)
)
print('Máscara booleana construida.')
print(f'Alumnos que cumplen las condiciones: {mascara.sum()}')
```

```python
df_destacados = df_notas[mascara]
print('DataFrame filtrado guardado en df_destacados.')
```

```python
print('=== ALUMNOS DESTACADOS EN PROGRAMACIÓN ===')
df_destacados
```

# Ejercicio 5: El Desafío Final - Estadística Vectorizada

```python
df_redes = df_notas[df_notas['Carrera'] == 'Redes']
print(f'Alumnos de Redes: {len(df_redes)}')
```
