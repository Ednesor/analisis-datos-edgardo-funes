# Ejercicio 1: Creación y Atributos

```python
import numpy

lista = numpy.array([1, 2, 3, 4, 5])
print(f"Dimensión: {lista.ndim}")
print(f"Forma: {lista.shape}")
print(f"tipo: {lista.dtype}")
lista_float = lista.astype(float)
print(f"tipo: {lista_float.dtype}")
```

# Ejercicio 4: El Simulador de Inventario (Integración)

```python
import numpy as np
cantidadSucursales = int(input("Ingrese la cantidad de sucursales: "))
cantidadProductos = int(input("Ingrese la cantidad de productos: "))

matriz = np.zeros((cantidadSucursales, cantidadProductos), dtype=int)
print(matriz)
print("Base de datos de inventario inicializada correctamente")
```

# Ejercicio 5: Redimensionando la realidad (Reshape)

```python
import numpy as np

array = np.arange(1, 13)
print(array.ndim)

newArray = np.reshape(array, (3, 4))

print(newArray.ndim)
```
