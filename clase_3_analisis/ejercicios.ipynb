# Ejercicio 1: Slicing y Sub-matrices

```python
import numpy as np
lista = np.arange(1, 26).reshape(5, -1)

matrizCentral = lista[1:4, 1:4]

ultimaColumna = lista[0:6, 4:5]

print(f"Lista: {lista} \n")
print(f"Matriz central: {matrizCentral} \n")
print(f"Utlima columna: {ultimaColumna} \n")
```

# Ejercicio 2: Estadísticas con Ejes (Axis)

```python
import numpy as np

ventas = np.array([
  [1500, 2000, 1800, 2200],
  [1100, 1300, 1200, 1500],
  [3000, 3100, 2900, 3500]
])

totalVentas = ventas.sum()

promedioSemanal = ventas.mean(axis=0)

maximoSucursal = ventas.max(axis=1)

print(f"Total ventas: {totalVentas}\n")
print(f"Promedio semanal: {promedioSemanal}\n")
print(f"Maxipo por sucursal: {maximoSucursal}\n")
```

# Ejercicio 3: El poder de la Máscara Booleana

```python
import numpy as np

array = np.arange(1, 21)

mascara = array % 2 == 0

numerosPares = array[mascara]

print(f"Mascara: {mascara}")
print(f"Numeros pares: {numerosPares}")
```

# Ejercicio 4: Condiciones Múltiples (El Analista de RRHH)

```python
import numpy as np

edades = np.array([22, 55, 34, 19, 45, 61, 28, 30, 42])

mascara = (edades >= 30) & (edades <=50)
mascaraExtremos = (edades < 25) | (edades > 55)

trabajadoresFiltrados = edades[mascara]
trabajadoresFiltradosExtremos = edades[mascaraExtremos]

print(f"Trabajadores entre 30 y 50 años: {trabajadoresFiltrados}")
print(f"Trabajadores menores de 25 O mayores de 55 años: {trabajadoresFiltradosExtremos}")
```

# Ejercicio 5: Modificación Condicional

```python
import numpy as np

matriz = np.zeros((4, 4), dtype=int)
print(f"Matriz original: \n{matriz}\n")

matriz[1:3, 1:3] = 9
print(f"Matriz actualizada: \n{matriz}\n")

```

# Ejercicio 6: Desafío Integrador "El Profesor Robot"

```python
import numpy as np

calificaciones = np.random.randint(1, 11, 50)
print(f"Calificaciones: {calificaciones}")

promedio = calificaciones.mean()
print(f"Promedio: {promedio}")

mascaraDesaprobados = calificaciones < 4

totalDesaprobados = calificaciones[mascaraDesaprobados].size
print(f"Total desaprobados: {totalDesaprobados}")

calificaciones[mascaraDesaprobados] = 4

print(f"Calificaciones actualizdas: {calificaciones}")
```
